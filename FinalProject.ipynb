### Importing Data, Cleaning, Merging, and Creating Negative Samples

```python
# Load CSVs
firms = pd.read_csv("Data/fire_archive_M-C61_609028.csv")
soil = pd.read_csv("Data/soil_data.csv")

# Round coordinates for joining consistency
firms['lat_rounded'] = firms['latitude'].round(2)
firms['lon_rounded'] = firms['longitude'].round(2)
soil['lat_rounded'] = soil['lat'].round(2)
soil['lon_rounded'] = soil['lon'].round(2)

# Merge FIRMS and Soil data
merged_pos = pd.merge(firms, soil, on=['lat_rounded', 'lon_rounded'], how='inner')

# Drop rows that didn't match cleanly (already handled by inner join)
# Label positive fire examples
merged_pos['fire_occurred'] = 1

# Generate negative fire samples (no fire occurred)

# Define bounding box from FIRMS data
lat_min, lat_max = firms['latitude'].min(), firms['latitude'].max()
lon_min, lon_max = firms['longitude'].min(), firms['longitude'].max()

# Sample locations from soil data (assumed complete and valid)
n_neg = len(merged_pos)
neg_locations = soil.sample(n=n_neg, replace=True, random_state=42)

# Add random dates
neg_locations['acq_date'] = np.random.choice(firms['acq_date'], size=n_neg)

# Create negative samples
neg_locations['fire_occurred'] = 0

# Make columns consistent
columns_to_keep = merged_pos.columns.intersection(neg_locations.columns)
merged_neg = neg_locations[columns_to_keep]

# Combine positive and negative examples

final_df = pd.concat([merged_pos, merged_neg], ignore_index=True)

# Drop helper columns
final_df.drop(columns=['lat_rounded', 'lon_rounded'], inplace=True)

# Preview result
print(final_df.shape)
print(final_df['fire_occurred'].value_counts())
print(final_df.head())
```
